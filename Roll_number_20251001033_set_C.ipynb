```python
import math

a = kc
b = -(kc + 1)
c = 0.25 * kc

discriminant = b**2 - 4 * a * c
sqrt_discriminant = math.sqrt(discriminant)

root1 = (-b - sqrt_discriminant) / (2 * a)
root2 = (-b + sqrt_discriminant) / (2 * a)

print("a =", a)
print("b =", b)
print("c =", c)
print("Discriminant =", discriminant)
print("Root 1 =", root1)
print("Root 2 =", root2)
```

```python
#1
C0 = 1.0       # Initial concentration (mol/L)
k = 0.12       # Rate constant (min^-1)
delta_t = 0.5  # Time step (min)
C = C0

print("Time (min)\tConcentration (mol/L)\tConversion (%)")
print(f"{0.0:.1f}\t\t{C:.4f}\t\t\t{0.0:.2f}")

for step in range(1, 11):  # 5 minutes / 0.5-minute steps
    C = C - k * C * delta_t
    time = step * delta_t
    conversion = (C0 - C) / C0 * 100

    print(f"{time:.1f}\t\t{C:.4f}\t\t\t{conversion:.2f}")
```

```python
#part A (2)
import math

C0 = 1.0
k = 0.12
delta_t = 0.5
time_end = 5.0

# Euler simulation
C_euler = C0
for _ in range(int(time_end / delta_t)):
    C_euler = C_euler - k * C_euler * delta_t

# Exact solution
C_exact = C0 * math.exp(-k * time_end)

absolute_error = abs(C_exact - C_euler)

print(f"Euler concentration: {C_euler:.6f} mol/L")
print(f"Exact concentration: {C_exact:.6f} mol/L")
print(f"Absolute error:      {absolute_error:.6f} mol/L")
```

```python
#question 2 part A part 3
C0 = 1.0
k = 0.12
delta_t = 0.5

C = C0
time = 0.0
conversion = 0.0

while conversion < 70 and time < 100:
    C = C - k * C * delta_t
    time += delta_t
    conversion = (C0 - C) / C0 * 100

print(f"Time needed: {time:.1f} min")
print(f"Final concentration: {C:.6f} mol/L")
print(f"Final conversion: {conversion:.2f}%")
```

```python
# question 3 part A 
def factorial(n):
    if n < 0:
        raise ValueError("Factorial is not defined for negative numbers.")
    if n == 0:
        return 1
    return n * factorial(n - 1)


def combination(n, r):
    if n < 0 or r < 0:
        raise ValueError("n and r must be non-negative.")
    if r > n:
        raise ValueError("r cannot be greater than n.")

    return factorial(n) // (factorial(r) * factorial(n - r))


try:
    n = int(input("Enter the number of candidates (n): "))
    r = int(input("Enter how many candidates to blend (r): "))

    print(f"\nValues entered: n = {n}, r = {r}")

    result = combination(n, r)
    print(f"Number of possible combinations: {result}")

except ValueError as error:
    print("Invalid input:", error)
```

```python
#question 3 part B
def fib_iter(n):
    """Return a list of the first n Fibonacci numbers."""
    if n < 0:
        raise ValueError("n must be non-negative.")

    numbers = []
    a, b = 0, 1

    for _ in range(n):
        numbers.append(a)
        a, b = b, a + b

    return numbers


def fib_rec(k):
    """Return the k-th Fibonacci number (F0 = 0, F1 = 1)."""
    if k < 0:
        raise ValueError("k must be non-negative.")
    if k <= 1:
        return k
    return fib_rec(k - 1) + fib_rec(k - 2)


try:
    n = int(input("How many readings? "))
    k = int(input("Enter k to find the k-th Fibonacci number: "))

    fibonacci_numbers = fib_iter(n)

    print(f"\n{'Reading No.':<15}{'Fibonacci tag':<18}{'Time (min)':<12}")
    print("-" * 45)

    for reading_no, fib_number in enumerate(fibonacci_numbers, start=1):
        time = fib_number * 3
        print(f"{reading_no:<15}{fib_number:<18}{time:<12}")

    print(f"\nF{k} = {fib_rec(k)}")

except ValueError as error:
    print("Invalid input:", error)
```

```python
data = {
    "time_min": [0, 10, 20, 30, 40, 50],
    "Run1_mol_l": [0.80, 0.54, 0.36, 0.24, 0.16, 0.11],
    "Run2_mol_L": [0.80, 0.44, 0.24, 0.13, 0.07, 0.04]
}

Kinetics = pd.DataFrame(data)

# Save the data files
Kinetics.to_csv("Kinetics.csv", index=False)
Kinetics.to_json("Kinetics.json", orient="records", indent=4)

# Read the CSV file back
Kinetics_read = pd.read_csv("Kinetics.csv")

print("Shape:", Kinetics_read.shape)
print(Kinetics_read)
```

```python
#question 4 part d
import numpy as np
import matplotlib.pyplot as plt

volume_L = np.array([2, 3, 4, 5, 6])
pressure_bar = np.array([3.01, 1.98, 1.52, 1.19, 1.01])

# Calculate the mean of P * V
k = np.mean(pressure_bar * volume_L)
print(f"Mean value of P × V, k = {k:.2f} bar·L")

# Generate points for the reference curve P = k / V
volume_curve = np.linspace(volume_L.min(), volume_L.max(), 200)
pressure_curve = k / volume_curve

plt.scatter(volume_L, pressure_bar, label="Measured data")
plt.plot(volume_curve, pressure_curve, color="red", label="Reference curve: P = k/V")

plt.title("Pressure vs. Volume at Constant Temperature")
plt.xlabel("Volume (L)")
plt.ylabel("Pressure (bar)")
plt.legend()
plt.grid(True)
plt.show()
```
