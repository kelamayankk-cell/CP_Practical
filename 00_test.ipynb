```python
# get marks of a student and give a grade on the marks 
# 95 and above - grade A, 85 and above - grade B, 75 and above - grade C, 65 and above - grade D, 55 and above - grade E 
marks = int(input('Enter marks of student: '))  
if marks >= 95:
    print('Grade A')
elif marks >= 85:
    print('Grade B')
elif marks >= 75:
    print('Grade C')
elif marks >= 65:
    print('Grade D')
elif marks >= 55:
    print('Grade E')
else:
    print('Grade F')    
```
